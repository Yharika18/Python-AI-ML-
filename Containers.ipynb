## Functions in list

```python
# Functions in List
l1=[10,9,18,20,9,46,18,25,18]
print(len(l1))
print(l1.count(18))
print(l1.index(9)) 
# append function add the items at the end of the list
l1.append(45)
print(l1)
# Insert function add the item at the given index
l1.insert(5,'Hari')
print(l1)
# Extend function is used to add the all items of one list into another lists
l2=['Python',40,65,'Code',85]
l1.extend(l2)
print(l1)
```

# Delete/Remove elements in list 

```python
# Remove function is used to remove specific element in the list
l1.remove(18)
print(l1)

# pop function is used to pop the last value in the list
l2.pop()
print(l2)
```

```python
# reverse function is used to reverse the elements in a list
print(l2)
l2.reverse()
print(l2)

# sort function is used to sort the elemnts into the list
l3=[80,70,40,10,60,50]
l3.sort()
print(l3)
```

# List Comprehensions
creating list objects from any iterable objects based on some condition

syntax: list=[expression for item in list  if condition] 

```python
w1=["Python","Machine Learning","GenAI","Code"]
r=[w[0] for  w in w1] 
print(r)
```

```python
words=["Python","Machine Learning","GenAI","Code"] 
vowels = [ch for word in words for ch in word if ch.lower() in "aeiou"]

print(vowels)
```

# Set
1. If we want to represent a group of unique values as a single entity then we should go for set. 
2. Duplicates are not allowed. 
3. Insertion order is not preserved.But we can sort the elements. 
4. Indexing and slicing not allowed for the set. 
5. Heterogeneous elements are allowed. 
6. Set objects are mutable i.e once we creates set object we can perform any changes in that object based on our requirement. 
7. We can represent set elements within curly braces and with comma seperation 

# Creating the sets

```python
# method1 creating
s1={10,20,30,40}   
print(s1) 
# method2 creating
l = [10,20,30,40,10,20,10]   
s2=set(l)
print(s2)

# method3 creating
s=set(range(5))   
print(s)
```

#### In Sets we can't do indexing and slicing

# Functions in Sets

```python
# add function in sets
s1.add(80)
print(s1)
```

```python
#update
s={10,20,30}   
l=[40,50,60,10]   
s.update(l,range(5))   
print(s)
```

```python
# copy
s={10,20,30} 
s1=s.copy() 
print(s1) 

# pop
s={40,10,30,20}   
print(s)   
print(s.pop())   
print(s) 

#remove
s={40,10,30,20} 
s.remove(30) 
print(s) 

# It removes specific element we provided
s={10,20,30} 
s.discard(10) 
print(s) 
```

```python
# union return all elements present in the both sets
x={10,20,30,40} 
y={30,40,50,60} 
print(x.union(y))
```

```python
# intersection return only common elements present in both sets
x={10,20,30,40} 
y={30,40,50,60} 
print(x.intersection(y)) 
```

```python
# difference is it return the elements present in x set but not y set
x={10,20,30,40} 
y={30,40,50,60} 
print(x.difference(y))
```

 # Dictionaries
1. Duplicate keys are not allowed but values can be duplicated. 
2. Hetrogeneous objects are allowed for both key and values. 
3. insertion order is not preserved 
4. Dictionaries are mutable 
5. Dictionaries are dynamic 
6. indexing and slicing concepts are not applicable 

```python
# creating Dictionary
d={}
d1=dict()
d[100]="durga" 
d[200]="ravi" 
d[300]="shiva" 
print(d)
```

# Accessing from Dictionary

```python
# we can access data by using keys
print(d)
print(d[200])
```

```python
# call items from dictionary
my_dict={'key1':123,'key2':456,'key3': ['Python', 'Java', 'SQL']}
# call an index on that value
print(my_dict['key2'])
print(my_dict['key3'][0])
# can then even call methods on that value
print(my_dict['key3'][0].upper())
```

# Nesting dictionaries

```python
d={'key1':{'nestkey':{'subnestkey':123}}}
print(d)
print(d['key1']['nestkey'])
```

# Dictionary Comprehension

```python
squares={x:x*x for x in range(1,6)}   
print(squares)
```
