## 1. What is Generative AI?

- **Generative AI** creates new content such as text, code, images, summaries, answers, and recommendations.
- It learns patterns from large datasets and generates output based on user prompts.
- In LLM applications, GenAI is commonly used for chatbots, document Q&A, summarization, coding assistants, recommendation systems, and automation.
- Core building blocks: **LLM, prompt, context, embeddings, vector database, retriever, tools, memory, and agent.**

## 2. What is LangChain?

- **LangChain** is a framework used to build LLM-powered applications.
- It helps connect LLMs with prompts, documents, APIs, databases, tools, memory, and agents.
- It simplifies building applications like RAG chatbots, AI agents, document Q&A systems, and automation workflows.

**Simple meaning:** LangChain helps developers connect an LLM with external data and actions.

## 3. What is LangGraph?

- **LangGraph** is used to build more controlled, stateful, multi-step AI workflows and agents.
- It represents an AI workflow as a **graph** with nodes and edges.
- A node can be an LLM call, tool call, retriever, condition, human approval step, or memory update.
- It is useful when the application needs loops, branching, retries, state management, or human-in-the-loop.

**Simple meaning:** LangGraph gives more control over complex agent workflows.

## 4. What is LangSmith?

- **LangSmith** is used for tracing, debugging, testing, monitoring, and evaluating LLM applications.
- It shows what prompt was sent, what context was retrieved, what tool was called, how many tokens were used, and where failures happened.
- It helps improve quality, reduce latency, and compare different prompts/models/retrievers.

**Simple meaning:** LangSmith is like an observability and evaluation tool for LLM apps.

## 6. What is RAG?

**RAG = Retrieval-Augmented Generation**

- RAG allows an LLM to answer using external data instead of relying only on its training knowledge.
- It retrieves relevant information from documents, databases, or knowledge bases and passes that context to the LLM.
- It reduces hallucination and improves domain-specific accuracy.

**RAG flow:**
1. User asks a question.
2. Convert the question into an embedding.
3. Search similar chunks in a vector database.
4. Retrieve relevant context.
5. Send context + question to the LLM.
6. Generate final answer.

## 7. Types of RAG

- **Naive RAG:** Basic chunking → embedding → vector search → answer generation.
- **Advanced RAG:** Adds better chunking, metadata filtering, reranking, query rewriting, and hybrid search.
- **Agentic RAG:** An agent decides when to retrieve, which tool/database to use, and whether more search is needed.
- **Corrective RAG:** Checks whether retrieved context is useful; if not, it searches again or rewrites the query.
- **Hybrid RAG:** Combines vector search with keyword/BM25 search for better recall.
- **Graph RAG:** Uses relationships between entities and documents, often useful for complex enterprise knowledge.

## 8. What does “Chain” mean in LangChain?

A **chain** means connecting multiple steps together.

Example:
- Step 1: Take user question.
- Step 2: Retrieve related documents.
- Step 3: Format prompt.
- Step 4: Send prompt to LLM.
- Step 5: Return final answer.

**Simple meaning:** A chain is a pipeline of components where output from one step becomes input to the next step.

## 9. Components of RAG

- **Data source:** PDF, website, database, CSV, documentation, CRM, etc.
- **Document loader:** Loads data from the source.
- **Text splitter/chunker:** Breaks large documents into smaller chunks.
- **Embedding model:** Converts text into numerical vectors.
- **Vector database:** Stores embeddings and retrieves similar chunks.
- **Retriever:** Searches and returns relevant chunks.
- **Reranker:** Reorders retrieved chunks by relevance.
- **Prompt template:** Combines user question and retrieved context.
- **LLM:** Generates the final answer.
- **Evaluator/monitoring:** Checks answer quality, latency, cost, and hallucination.

## 10. What is Chunking?

- **Chunking** means splitting large text into smaller pieces before embedding.
- LLMs and embedding models have token limits, so large documents cannot be processed as one block.
- Good chunking improves retrieval accuracy.
- Bad chunking loses context and gives weak answers.

**Why chunking is needed:**
- Handles token limits.
- Improves search quality.
- Reduces cost.
- Improves latency.
- Keeps retrieved context focused.

## 11. Types of Chunking Techniques

| Type | Meaning | Best Use Case | Risk |
|---|---|---|---|
| Fixed-size chunking | Split by fixed token/character size | Simple documents | May cut meaning in the middle |
| Recursive chunking | Split by paragraphs, sentences, then words | General RAG systems | Needs tuning |
| Semantic chunking | Split based on meaning/topic change | Complex documents | More compute cost |
| Sentence-based chunking | Split by complete sentences | FAQs, policies | Chunks may be too small |
| Paragraph/section chunking | Split by headings or sections | Reports, manuals, legal docs | Large sections may exceed token limit |
| Sliding-window chunking | Chunks overlap with previous chunks | When context continuity matters | More storage and cost |

## 12. Chunking Optimization

- Use **smaller chunks** for precise Q&A.
- Use **larger chunks** when more context is required.
- Use **overlap** to avoid losing meaning between chunks.
- Too much overlap increases storage, token cost, and retrieval noise.
- Add **metadata** like document name, page number, section, date, and category.

**Good default:** 500–1,000 tokens per chunk with 10–20% overlap. Tune based on your data.

## 13. What are Vector Embeddings?

- **Vector embeddings** are numerical representations of text.
- Similar meanings are placed closer together in vector space.
- Example: “leave policy” and “vacation policy” may be close even if the words are different.
- Embeddings help the system perform semantic search instead of only keyword search.

````python
# Simple conceptual example of an embedding
text = "What is the company leave policy?"
embedding_vector = [0.12, -0.45, 0.88, 0.03]  # Real embeddings have hundreds or thousands of dimensions
print(embedding_vector)
````

## 14. How Data is Stored as Vector Embeddings

1. Load documents.
2. Split documents into chunks.
3. Convert each chunk into an embedding vector.
4. Store the vector in a vector database.
5. Store metadata along with the vector.
6. During search, convert the user query into a vector.
7. Find the most similar stored vectors.

**Stored record usually contains:**
- Chunk text
- Embedding vector
- Document ID
- Page number
- Source name
- Metadata
- Timestamp/version

## 15. Types / Components of Embeddings

- **Text embeddings:** Used for document search and Q&A.
- **Code embeddings:** Used for code search and code assistants.
- **Image embeddings:** Used for image similarity search.
- **Multimodal embeddings:** Represent text, image, audio, or video together.
- **Dense embeddings:** Semantic vectors from neural models.
- **Sparse embeddings:** Keyword-style vectors, useful for exact word matching.
- **Hybrid embeddings/search:** Combines dense semantic search and sparse keyword search.

## 16. What is a Vector Database?

- A **Vector Database (VDB)** stores embeddings and performs similarity search.
- It helps retrieve the most relevant chunks for a user query.
- It uses algorithms like approximate nearest neighbor search to search quickly.

**Why VDB is used:**
- Fast semantic search.
- Handles large document collections.
- Supports metadata filtering.
- Supports scalable retrieval for RAG.

## 17. ChromaDB vs FAISS vs Pinecone

| Feature | ChromaDB | FAISS | Pinecone |
|---|---|---|---|
| Type | Open-source vector DB | Vector similarity library | Managed cloud vector DB |
| Best for | Local prototypes, small/medium apps | Fast local similarity search | Production-scale managed RAG |
| Hosting | Local/self-hosted | Local/self-managed | Fully managed cloud |
| Metadata filtering | Good | Limited/manual | Strong |
| Scalability | Moderate | High but self-managed | High and managed |
| Ease of use | Easy | Medium | Easy for production |
| Cost | Low/free | Low/free | Paid managed service |

**Interview answer:** Use ChromaDB for quick prototypes, FAISS for fast local vector search, and Pinecone when production scalability and managed infrastructure are needed.

## 18. What is Query Decomposition?

- **Query decomposition** breaks a complex user question into smaller sub-questions.
- It is useful when one question needs multiple facts or multiple retrieval steps.

Example:
Question: “Compare AWS Lambda and EC2 for cost, scalability, and use cases.”
Sub-questions:
1. What is AWS Lambda?
2. What is EC2?
3. Compare cost.
4. Compare scalability.
5. Compare best use cases.

**Why used:**
- Improves retrieval accuracy.
- Reduces missing information.
- Helps answer multi-part questions clearly.
- Useful in complex RAG and agentic workflows.

## 19. What is Reranking?

- **Reranking** reorders retrieved documents/chunks based on actual relevance to the query.
- First-stage retrieval may return many roughly similar chunks.
- Reranker selects the best chunks before sending context to the LLM.

**Why reranking is used:**
- Improves answer accuracy.
- Reduces irrelevant context.
- Reduces token usage.
- Improves final response quality.

## 20. Reranking Techniques

- **Cross-encoder reranking:** Query and document are scored together by a model. High quality but slower.
- **Embedding similarity reranking:** Reorder by cosine similarity score. Fast but less accurate.
- **LLM-based reranking:** Ask LLM to judge relevance. Flexible but costly and slower.
- **Metadata-based reranking:** Prioritize recent, trusted, or specific-source documents.
- **Hybrid reranking:** Combine semantic score, keyword score, metadata, and business rules.

**Optimization:** Retrieve top 20–50 chunks first, rerank them, then send only top 3–8 chunks to the LLM.

## 21. Reranking Optimization

- Do not rerank too many chunks; it increases latency.
- Use cheap retriever first, expensive reranker second.
- Use metadata filters before reranking.
- Cache common query results.
- Send only the most relevant chunks to the LLM.
- Measure accuracy vs latency tradeoff.

**Best practical flow:** Vector search top 30 → rerank top 30 → pass top 5 to LLM.

## 24. What is Memory in LLM Applications?

- **Memory** stores useful conversation or user information across turns.
- It helps the chatbot understand context instead of treating every message as new.

Example:
User: “My project is Retail Demand Forecasting.”
Later: “Explain the ML part.”
Memory helps the chatbot know which project the user means.

## 25. Types of Memory

| Memory Type | Meaning | Use Case | Risk |
|---|---|---|---|
| Buffer memory | Stores full conversation history | Short chats | Token usage grows fast |
| Window memory | Stores last N messages | Chatbots | Older context lost |
| Summary memory | Stores summarized conversation | Long chats | Summary may miss details |
| Entity memory | Stores facts about entities/users | Personalized assistants | Needs careful updating |
| Vector memory | Stores past interactions as embeddings | Long-term semantic recall | Retrieval quality must be tuned |
| Session memory | Stores context only during current session | Temporary workflows | Not persistent |
| Long-term memory | Stores durable user/project facts | Personal assistants | Privacy and data quality concerns |

## 26. Why Memory is Used

- Maintains conversation continuity.
- Reduces repeated questions.
- Personalizes responses.
- Helps multi-turn workflows.
- Reduces prompt size when summarized properly.

**But:** Memory must be optimized. Dumping full chat history into every prompt increases token cost and latency.

## 27. Memory Optimization

- Use **window memory** for short support chats.
- Use **summary memory** for long conversations.
- Use **vector memory** for long-term semantic recall.
- Store only useful facts, not every message.
- Remove duplicate and outdated memory.
- Summarize old context to reduce tokens.
- Use metadata like user ID, session ID, project ID, and timestamp.

**Best practice:** Recent messages + summarized history + retrieved long-term memory.

## 28. What is an Agent?

- An **agent** is an LLM-based system that can decide what action to take next.
- It can use tools, memory, planning, retrieval, and reasoning.
- Agents are useful when the task is not a simple one-step answer.

**Agent loop:**
1. Understand user goal.
2. Decide next step.
3. Use tool or retrieve data.
4. Observe result.
5. Continue or final answer.

**Simple meaning:** An agent is an LLM that can make decisions and use tools to complete tasks.

## 29. Agent vs Chain

| Chain | Agent |
|---|---|
| Fixed sequence of steps | Dynamic decision-making |
| Predictable flow | Chooses tools/actions based on need |
| Easier to debug | More flexible but harder to control |
| Lower latency | Can have higher latency |
| Best for standard workflows | Best for complex, open-ended tasks |

**Use chain when process is fixed. Use agent when the system must decide the next action.**

## 30. Token Limit, Token Count, API Calls, Latency

- **Token limit:** Maximum tokens a model can process in input + output.
- **Token count:** Number of tokens used in prompt, context, and response.
- **API calls:** Each model/tool/retriever/reranker call adds time and cost.
- **Latency:** Total time taken to generate final response.

**More tokens = more cost + more latency.**

**More API calls = more latency + more failure points.**

## 31. How to Prove Optimization

Track these metrics before and after optimization:

| Metric | Before | After | Goal |
|---|---:|---:|---|
| Average latency | 6 sec | 3 sec | Reduce response time |
| Input tokens | 8,000 | 3,500 | Reduce prompt size |
| Output tokens | 1,000 | 600 | Keep answer concise |
| API calls | 6 | 3 | Reduce cost and failure points |
| Retrieval precision | 65% | 85% | Improve context relevance |
| Hallucination rate | High | Low | Improve grounded answers |
| Cost per query | $0.08 | $0.03 | Reduce cost |

**Do not just say “optimized.” Prove it using numbers.**

## 32. Less Latency: How to Achieve It

- Reduce number of retrieved chunks.
- Use reranking only when needed.
- Cache frequent questions and embeddings.
- Use smaller/faster model for simple tasks.
- Use metadata filters before vector search.
- Limit conversation history.
- Use summary memory instead of full memory.
- Avoid unnecessary agent loops.
- Parallelize independent tool calls when possible.

**Main idea:** Fewer tokens + fewer calls + better retrieval = lower latency.

## 33. Why This Approach, Not Another?

### Why RAG instead of fine-tuning?
- RAG is better when knowledge changes often.
- RAG is cheaper and easier to update.
- Fine-tuning is better for behavior/style, not frequently changing facts.

### Why vector DB instead of normal SQL search?
- Vector DB understands meaning, not just exact keywords.
- SQL is better for structured data; vector DB is better for semantic document search.

### Why reranking?
- Vector search may return noisy chunks.
- Reranking improves relevance before LLM generation.

### Why LangGraph instead of only LangChain chains?
- LangGraph is better for complex workflows with state, branching, retries, and human approval.

### Why tools?
- Tools allow agents to access real-time external systems and take actions.

## 34. Practical RAG Design Decision Table

| Situation | Best Choice | Reason |
|---|---|---|
| Small prototype | ChromaDB | Easy local setup |
| Fast local search | FAISS | High-performance similarity search |
| Production managed app | Pinecone | Scalable and managed |
| Fixed workflow | Chain | Lower latency and easier debugging |
| Complex workflow | Agent/LangGraph | Dynamic tool selection and state |
| Long conversation | Summary + vector memory | Controls token usage |
| High noise retrieval | Reranking | Improves context quality |
| Complex question | Query decomposition | Better multi-step retrieval |
