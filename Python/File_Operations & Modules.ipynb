# Except Block
In except block to handle any type of exceptions. 
In default except block generally we can print normal error messages.

# finally block: 
 
1. It is not recommended to maintain clean up code(Resource Deallocating Code or 
Resource Releasing code) inside try block  because there is no guarentee for the execution 
of every statement inside try block always. 
 
2.  It is not recommended to maintain clean  up code inside except block, because if there 
is no exception then except block won't be executed. 


```python
try:
    f=open("testfile.txt","w")
    f.write("Test write statement")
finally:
    print("Always execute finally code blocks")
```

```python
def askint():
    try:
        val=int(input("Enter an Integer:"))
    except:
        print(" looks like you didnot enter integer!")
        try:
            val= int(input("Try again please enter integer"))
        except:
            print("handle it")
    finally:
        print("Finally, I excecuted!")
askint()
    
```

# Database Connectivity and Operations using python

- Database connectivity means creating a connection between a Python program and a database to perform operations like
inserting, retrieving, updating, and deleting data.
- CRUD Operations
- C - Create / Insert data
- R - Read / Fetch data
- U - Update existing data
- D - Delete data

```python
# Creating  a Table
import sqlite3

connection = sqlite3.connect("student.db")
cursor = connection.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    course TEXT
)
""")

connection.commit()
connection.close()

print("Table created successfully")
```

```python
# INserting Data Into Table
import sqlite3

connection = sqlite3.connect("student.db")
cursor = connection.cursor()

cursor.execute("""
INSERT INTO students (name, age, course)
VALUES ('Ravi', 22, 'Python')
""")

connection.commit()
connection.close()

print("Data inserted successfully")
```

```python
# Reading / Fetching Data
import sqlite3

connection = sqlite3.connect("student.db")
cursor = connection.cursor()

cursor.execute("SELECT * FROM students")

records = cursor.fetchall()

for row in records:
    print(row)

connection.close()
```

```python
# Updating Data
import sqlite3

connection = sqlite3.connect("student.db")
cursor = connection.cursor()

cursor.execute("""
UPDATE students
SET course = ?
WHERE name = ?
""", ("Data Science", "Ravi"))

connection.commit()
connection.close()

print("Data updated successfully")
```

```python
# Deleting Data
import sqlite3

connection = sqlite3.connect("student.db")
cursor = connection.cursor()

cursor.execute("""
DELETE FROM students
WHERE name = ?
""", ("Anita",))

connection.commit()
connection.close()

print("Data deleted successfully")
```
